# Question 1
Create a higher-order function that takes 3 parameters: A function and the two parameters that that function takes, and
flips the order of the parameters.

For example this: `(/) 6 2` returns `3`. But this: `flip' (/) 6 2` returns `0.3333333333`

```haskell
flip' :: (a -> a -> a) -> a -> a -> a
flip' f x y = f y x  -- not f (y x) !!! (y x) means "apply y as a function to x" — but y is just an a, not a function.
flip'(/) 6 2
```

# Question 2
Create the `uncurry'` function that converts a curried function to a function on pairs.

So this: `(+) 1 2` that returns `3` can be written as
`uncurry' (+) (1,2)` (with the two different arguments inside a pair).

```haskell
:t (+)
res = (+) 1 2
```

```haskell
-- f is a curried function: has type a -> a -> a
uncurry' :: (a -> a -> a) -> (a, a) -> a
uncurry' f  (x, y) = f x y
```

```haskell
res' = uncurry' (+) (1, 2)
res == res'
```

# Question 3
Create the `curry'` function that converts an uncurried function to a curried function.
So this: `fst (1,2)` that returns `1` can be written as
`curry' fst 1 2` (with the tuple converted into two different arguments).

```haskell
:t fst
res' = fst (1, 2)
```

```haskell
curry' :: ((a, b) -> a) -> a -> b -> a
curry' f x y = f (x, y)
```

```haskell
res'' = curry' fst 1 2
res' == res''
```

# Question 4
Use higher-order functions, partial application, and point-free style to create a function that checks if a word has an uppercase letter.
Start with using just higher-order functions and build from there. 

```haskell
:t any
import Data.Char  -- eh questo mica me lo ha spiegato
:t isUpper
:t fromEnum
fromEnum 'A'
fromEnum 'Z'
fromEnum 'a'
fromEnum 'z'
isUpper' :: Char -> Bool
isUpper' c = (a >= fromEnum 'A') && (a <= fromEnum 'Z') where  -- and is not syntactic sugar
    a = fromEnum c
```

```haskell
-- high-order function + partial application
hasWordUpper :: [Char] -> Bool
hasWordUpper s = any isUpper' s

-- point-free style
hasWordUpper'' :: [Char] -> Bool
hasWordUpper'' = any isUpper
```

```haskell
hasWordUpper "tacchini" == False
hasWordUpper "Coda" == True
hasWordUpper'' "tacchini" == hasWordUpper "tacchini"
hasWordUpper'' "Coda" == hasWordUpper "Coda"
```

# Question 5
Create the `count` function that takes a team ("Red", "Blue", or "Green") and returns the amount of votes the team has inside `votes`.

```haskell
votes :: [String]
votes = ["Red", "Blue", "Green", "Blue", "Blue", "Red"]
:t filter
-- filter == votes
areRed = filter (== "Red") votes
:t areRed
show areRed
length areRed
```

```haskell
count :: String -> Int
count t = length (filter (== t) votes)
count "Red" == 2
count "Blue" == 3
count "banana" == 0
```

# Question 6
Create a one-line function that filters `cars` by brand and then checks if there are any left.

```haskell
cars :: [(String, Int)]
cars = [("Toyota", 0), ("Nissan", 3), ("Ford", 1)]

null []
null [1]
```

```haskell
getCoupleNum :: (String, Int) -> Int
getCoupleNum t = case t of
    (b, n) -> n

snd ("Foo", 7)           == getCoupleNum ("Foo", 7)
```

```haskell
getBrandCars :: String -> [(String, Int)]
getBrandCars b = filter (\(brand, _) -> brand == b) cars

getBrandCars "Toyota"    == [("Toyota",0)]
getBrandCars "Foo"       == []

hasBrandCarLeft :: String -> Bool
hasBrandCarLeft b = case (getBrandCars b) of
    [] -> False
    [(bb, nn)] -> nn > 0
    _ -> False  -- TODO what about non-unique entries? e.g. cars' = [("Toyota", 1), ("Toyota", 3)]

hasBrandCarLeft "Miao"   == False
hasBrandCarLeft "Toyota" == False
hasBrandCarLeft "Nissan" == True
```

```haskell
hasBrandCarLeft' :: String -> Bool
hasBrandCarLeft' b = case (filter (\(brand, _) -> brand == b) cars) of
    [] -> False
    [(bb, nn)] -> nn > 0

hasBrandCarLeft "Miao"   == hasBrandCarLeft' "Miao"
hasBrandCarLeft "Toyota" == hasBrandCarLeft' "Toyota"
hasBrandCarLeft "Nissan" == hasBrandCarLeft' "Nissan"
```
