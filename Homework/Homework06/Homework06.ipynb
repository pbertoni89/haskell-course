# Question 3
Write a function called `concat'` that concatenates a list of lists.

```haskell
concat' :: [[a]] -> [a]
concat' [] = []
concat' (x : xs) = x ++ concat' xs  -- x is a list, we need to append two lists

a = concat  [[1, 2], [3], [4, 5, 6]]
b = concat' [[1, 2], [3], [4, 5, 6]]
a == b
```

# Question 4
Write a function called `zip'` that takes two lists and returns a list of
corresponding pairs (zips them) like this:

If one input list is shorter than the other, excess elements of the longer
list are discarded, even if one of the lists is infinite:

```haskell
zip [1, 2] ['a', 'b']
zip [1, 2] ['a']
zip [1]    ['a', 'b']
zip []     [1..]
zip [1..]  []

zip' :: [a] -> [b] -> [(a, b)]
zip' []       []       = []
zip' []       (_ : _)  = []
zip' (_ : _)  []       = []
zip' (x : xs) (y : ys) = [(x, y)] ++ zip' xs ys

zip'' :: [a] -> [b] -> [(a, b)]
zip'' (x : xs) (y : ys) = (x, y) : zip'' xs ys
zip''  _        _       = []

a   = zip   [1, 2] ['a', 'b']
a'  = zip'  [1, 2] ['a', 'b']
a'' = zip'' [1, 2] ['a', 'b']
a == a'
a == a''

b   = zip   [1, 2] ['a']
b'  = zip'  [1, 2] ['a']
b'' = zip'' [1, 2] ['a']
b == b'
b == b''
```

# Question 5
Create a function called `zipWith'` that generalises `zip'` by zipping with a
function given as the first argument, instead of a tupling function.

For example, `zipWith' (+)` is applied to two lists to produce the list of
corresponding sums:

```haskell
-- zipWith (,) xs ys == zip xs ys
-- zipWith f [x1,x2,x3] [y1,y2,y3] == [f x1 y1, f x2 y2, f x3 y3]
-- zipWith' f [x1,x2,x3..] [y1,y2,y3..] == [f x1 y1, f x2 y2, f x3 y3..]
zipWith (+) [1, 2, 3] [4, 5, 6]

zipWith' :: (a -> a -> a) -> [a] -> [a] -> [a]
zipWith' f (x : xs) (y : ys) = ((f) x y) : zipWith' f xs ys
zipWith'  _  _        _       = []

a  = zipWith  (+) [1, 2, 3] [4, 5, 6]
a' = zipWith' (+) [1, 2, 3] [4, 5, 6]
a == a'
```

# Question 6
Write a function called `takeWhile'` that takes a precate and a list and
returns the list up until an element that doesn't satisfy the predicate.

```haskell
takeWhile' :: (a -> Bool) -> [a] -> [a]
takeWhile' _  []            = []
takeWhile' f  (x:xs) | f x  = x : takeWhile' f xs
takeWhile' _  _             = []

l1 = [1,2,3,4,1,2,3,4]
l2 = [1,2,3]
a  = takeWhile  (< 3) l1
b  = takeWhile  (< 9) l2
c  = takeWhile  (< 0) l2
a' = takeWhile' (< 3) l1
b' = takeWhile' (< 9) l2
c' = takeWhile' (< 0) l2

a == a'
b == b'
c == c'
```

# Question 7 (More difficult)
Write a function that takes in an integer n, calculates the factorial n! and
returns a string in the form of 1*2* ... *n = n! where n! is the actual result.


```haskell
fac :: Int -> Int
-- The root issue with your original attempts was mixing guards and separate equations across definitions.
-- Using pattern matching on 0 instead of a guard sidesteps the problem entirely and is the most idiomatic
-- way to write factorial in Haskell.
--fac n | n <= 1    = 1
--fac n | otherwise = n * fac ((-) n 1)
fac 0 = 1
fac n = n * fac (n - 1)

fac 5 == 120

showFacList :: Int -> String
showFacList 0 = ""
showFacList n = showFacList (n - 1) ++ " " ++ show n

-- showFacL  3

showFac :: Int -> String
showFac n = showFacList n ++ " = " ++ show (fac n)

showFac 3

-- :t foldl  -- foldl :: (a -> b -> a) -> a -> [b] -> a
-- foldl with a starting accumulator of "" could append each number.
-- Can you write a lambda that takes the accumulator and the current number and produces the next string?
showFacL :: Int -> String
showFacL n = show (fac n) ++ foldl (\acc x -> acc ++ " " ++ show x) " =" [1..n]

reverse (showFacL 3)

-- see types !!!
-- In foldl the lambda is \accumulator element -> ...
-- In foldr the lambda is \element accumulator -> ...

showFacR :: Int -> String
showFacR n = foldr (\x acc -> show x ++ " " ++ acc) "" [1..n] ++ "= " ++ show (fac n)
showFacR 3
```

# Question 8
Below you have defined some beer prices in `bevogBeerPrices` and your order list in
`orderList` + the `deliveryCost`. Write a function that takes in an order and calculates
the cost including delivery. Assume that the two lists have the beers in the same order.

```haskell
bevogBeerPrices :: [(String, Double)]
bevogBeerPrices =
  [ ("Tak", 6.00),
    ("Kramah", 7.00),
    ("Ond", 8.50),
    ("Baja", 7.50)
  ]

orderList :: [(String, Double)]
orderList =
  [ ("Tak", 5),
    ("Kramah", 4),
    ("Ond", 7)
  ]

deliveryCost :: Double
deliveryCost = 8.50
```

```haskell
getTup :: String -> [(String, Double)] -> [(String, Double)]
getTup b l = filter (\(brand, _) -> brand == b) l

getSnd :: String -> [(String, Double)] -> Double
getSnd b l = case getTup b l of
    [] -> 0
    (x:_) -> snd x

getPrice :: String -> Double
getPrice b = getSnd b bevogBeerPrices

getNum :: String -> Double
getNum b = getSnd b orderList

-- getPrice "Tak"
-- getNum "Tak"

getOrderCost :: String -> Double
getOrderCost b = getPrice b * getNum b + deliveryCost

getOrderCost "Tak"
```
